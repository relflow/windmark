```python
# Initialize the list of dictionaries
data = [
    {'a': 1, 'b': 2, 'c': 3},
    {'a': 4, 'b': 5, 'c': 6},
    {'a': 7, 'b': 8, 'c': 9}
]

# Create a dictionary of lists using a dictionary comprehension
result = {key: [d[key] for d in data] for key in data[0]}

print(result)
```
